```python
df['Payment Method'].unique()
```

```python
yes_no_colums=['Senior Citizen','Partner','Dependents','Phone Service','Multiple Lines','Online Security', 'Online Backup', 'Device Protection', 
    'Tech Support', 'Streaming TV', 'Streaming Movies','Paperless Billing']
```

```python
for column in yes_no_colums:
    df[column]=df[column].map({'Yes':1,"No":0})
```

```python
for column in yes_no_colums:
    print(column, df[column].unique())
```

```python
df.info()
```

```python
ohe=OneHotEncoder(drop='first',sparse_output=False)
```

```python
encoded_array=ohe.fit_transform(df[['Payment Method','Internet Service']])
```

```python
new_column_names = ohe.get_feature_names_out(['Payment Method', 'Internet Service'])
```

```python
encoded_df = pd.DataFrame(encoded_array, columns=new_column_names, index=df.index)
```

```python
df=df.drop(columns=
           ['Payment Method','Internet Service']).join(encoded_df)
```
